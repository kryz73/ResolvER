```python
# Step 5: Diagnostics & Error Analysis
import json
import pandas as pd
from IPython.display import display

log_dir = Path("/kaggle/working/logs/kaggle_run")
summary_file = log_dir / "run_summary.json"

if summary_file.exists():
    with open(summary_file) as f:
        print("=== EXPERIMENT RUN SUMMARY ===")
        print(json.dumps(json.load(f), indent=2))

feat_file = log_dir / "feature_importance.csv"
if feat_file.exists():
    print("\n=== TOP 15 DRIVING FEATURES ===")
    display(pd.read_csv(feat_file).head(15))

fp_file = log_dir / "error_analysis_false_positives.csv"
if fp_file.exists():
    print("\n=== TOP FALSE POSITIVES ===")
    display(pd.read_csv(fp_file).head(10))

fn_file = log_dir / "error_analysis_false_negatives.csv"
if fn_file.exists():
    print("\n=== TOP FALSE NEGATIVES (MISSED MATCHES) ===")
    display(pd.read_csv(fn_file).head(10))
```
